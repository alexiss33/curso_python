**Listas**

Las listas son colecciones ordenadas y modificables (mutables) de elementos. Permiten duplicados y pueden contener datos de diferentes tipos, Se escriben con corchetes **[]**.

Su principal uso es para almacenar secuencias de elementos donde el orden importa y necesitas modificar los datos (añadir o borrar elementos)


```python
# Crear una lista de códigos de impuestos a fiscalizar
impuestos_activos = ["IVA", "Renta", "Retenciones", "Consumo"]
print(f"Lista de impuestos en campaña actual: {impuestos_activos}")

# Agregar un nuevo impuesto al plan de control (Método append)
impuestos_activos.append("Patrimonio")

# Modificar un elemento por su índice (Las listas son mutables)
# Corregimos el índice 2 para ser más específicos con el tipo de retención
impuestos_activos[2] = "Retención en la Fuente" 

# Acceder a elementos mediante Slicing (Rebanado de listas)
print(f"Impuesto prioritario (Índice 0): {impuestos_activos[0]}")
print(f"Nuevos impuestos añadidos al plan: {impuestos_activos[-2:]}")
print(f"Lista final de fiscalización: {impuestos_activos}\n")
```

**Diccionarios**

Son colecciones que almacenan datos en pares de **clave-valor**. No tienen un orden fijo tradicional, las claves son únicas y los valores pueden repetirse. Se escriben con llaves **{}**.

Generalmente se las utiliza para buscar información rápidamente usando un identificador único (como el RUC o el secuencial de una factura).

Los diccionarios son perfectos para modelar el expediente de un contribuyente. Permiten asociar una **clave única (como el RUC)** a sus atributos (Nombre, Estado Tributario, Deuda Pendiente).

```python
#print("=== 2. CLASE: DICCIONARIOS (Expediente de Contribuyentes) ===")

# Crear un diccionario con el perfil de riesgo de un contribuyente
contribuyente = {
    "id_fiscal": "900123456-7",
    "razon_social": "Inversiones Alianza SAS",
    "regimen": "General",
    "marca_omiso": False,
    "deuda_usd": 4500.00
}
print(f"Expediente inicial del contribuyente: {contribuyente}")

# Acceder a un valor específico mediante su llave (Consulta rápida)
print(f"Consultando deuda de {contribuyente['razon_social']}: ${contribuyente['deuda_usd']}")

# Actualizar datos: El área de cobranzas detecta una omisión y actualiza la deuda
contribuyente["marca_omiso"] = True
contribuyente["deuda_usd"] += 1500.00  # Sumamos una nueva sanción

# Consultar campos disponibles para auditoría (Llaves y valores)
print(f"Campos disponibles en la consulta: {list(contribuyente.keys())}")
print(f"Expediente actualizado en base de datos: {contribuyente}\n")
```

**Ejercicio en clase**

```python
usuario = {
    "nombre": "Carlos",
    "edad": 28,
    "telefono": None,
    "email": "carlos@email.com",
    "direccion": None
}

# 1. Identificar las claves con valores None
claves_perdidas = [clave for clave, valor in usuario.items() if valor is None]

# 2. Filtrar el diccionario original conservando solo lo que no sea None
usuario_limpio = {clave: valor for clave, valor in usuario.items() if valor is not None}

print(f"Campos vacíos detectados: {claves_perdidas}")
print(f"Diccionario limpio: {usuario_limpio}")
```

**Tuplas**

Son colecciones ordenadas pero **inmutables** (no se pueden modificar después de crearlas). Se escriben con paréntesis **()**. Al no cambiar, son más seguras y rápidas en memoria.

Son útiles para agrupar datos que nunca deben cambiar (como coordenadas geográficas de latitud y longitud).


En el ámbito tributario las usamos para datos que BAJO NINGUNA CIRCUNSTANCIA deben ser alterados durante la ejecución de un programa, como constantes legales, **tarifas fijas** de un año fiscal o alguna constante normativa.

```python
# Definir las tarifas fijas de IVA vigentes (por ejemplo: Exento, Reducido, General)
tarifas_iva_2026 = (0.0, 0.05, 0.15,0.12)
print(f"Tarifas oficiales de IVA (Inmutables): {tarifas_iva_2026}")

# Acceso por índice
print(f"Tarifa general de IVA: {tarifas_iva_2026[2] * 100}%")

# Desempaquetado de tuplas (Unpacking) para asignar a variables de cálculo
exento, reducido, general, antigua= tarifas_iva_2026
print(f"Variables asignadas -> Reducido: {reducido * 100}%, General: {general * 100}%")
print(f"Variables asignadas -> antigua: {antigua + 0.03}")

# Nota didáctica:
print("Nota fiscal: Si intentáramos cambiar tarifas_iva_2026[2] = 0.21, Python generará")
print("un error protegido para evitar la alteración fraudulenta o accidental de la norma.\n")
```

**DataFrame**

Un DataFrame en Python es una **estructura de datos tabular bidimensional**, de tamaño variable y con etiquetas en filas y columnas, similar a una hoja de cálculo de Excel o a una tabla de una base de datos SQL.

```python
import pandas as pd
import numpy as np

df_ventas = pd.DataFrame({
    'ID_Producto': [101, 102, 103, 104, 105, 106],
    'Categoria': ['Electrónica', 'Hogar', None, 'Juguetes', 'Electrónica', 'Hogar'],
    'Precio': [250.0, np.nan, 15.5, 40.0, np.nan, 85.0],
    'Unidades_Vendidas': [12, 5, 80, np.nan, 3, 14]
})

# 1. Total de nulos por columna
total_nulos = df_ventas.isnull().sum()

# 2. Porcentaje de nulos por columna
porcentaje_nulos = (df_ventas.isnull().sum() / len(df_ventas)) * 100

# Presentamos el diagnóstico combinado
diagnostico = pd.DataFrame({'Nulos': total_nulos, 'Porcentaje (%)': porcentaje_nulos})
print(diagnostico)
```

**Ejercicio con duplicados en dataframes**

```python
import pandas as pd

# Datos tributarios simulados (Facturas)
datos_fiscales = {
    'id_factura': ['FAC-001', 'FAC-002', 'FAC-003', 'FAC-001', 'FAC-004', 'FAC-005'],
    'nro_emisor': ['900123456-1', '800987654-2', '900123456-1', '900123456-1', '800987654-2', '900123456-1'],
    'monto_iva': [1900, 5000, 1200, 1900, 5000, 1900],
    'total': [11900, 31300, 7520, 11900, 31300, 11900]
}

df_tributario = pd.DataFrame(datos_fiscales)
print("DataFrame Original:")
print(df_tributario)
```

```python
# 1. Identificar cuáles filas son duplicadas (mantiene la primera aparición como única)
filas_duplicadas = df_tributario[df_tributario.duplicated()]

print("\n--- Facturas idénticas (Duplicados absolutos) ---")
print(filas_duplicadas)
```

El siguiente ejercicio es una variación de la anterior realizando la busqueda de duplicados a nivel de una sola columna **id_factura**

```python
# Detectar si un ID de factura se repite, sin importar si los montos son diferentes
facturas_id_repetido = df_tributario[df_tributario.duplicated(subset=['id_factura'], keep=False)]

print("\n--- Alerta: Facturas con el mismo ID registrado ---")
print(facturas_id_repetido)
```

Ahora se muestran las frecuencias que tienen los valores del dataframe

```python
# Agrupar y contar las frecuencias de combinaciones sospechosas
reporte_duplicados = df_tributario.groupby(['nro_emisor', 'total']).size().reset_index(name='frecuencia')

# Filtrar solo aquellas combinaciones que aparecen más de una vez
patrones_sospechosos = reporte_duplicados[reporte_duplicados['frecuencia'] > 1]

print("\n--- Reporte: Misma empresa emitiendo montos idénticos ---")
print(patrones_sospechosos)
```

```python
reporte_duplicados2 = df_tributario.groupby(['nro_emisor', 'total']).size()
```

* **groupby(['nro_emisor', 'total']):** Divide el DataFrame en grupos basados en los valores de las columnas indicadas.
* **.size():** Cuenta cuántas filas o registros existen dentro de cada uno de los grupos creados por el groupby
* **.reset_index(name='frecuencia'):** Devuelve el objeto a un formato de DataFrame estándar (con filas y columnas tradicionales) y le asigna un nombre a la columna que contiene el conteo.
* **df[df['frecuencia'] > 1]:** Aplica una máscara booleana para conservar únicamente las filas que cumplen la condición matemática.

```python
# TEMA: Funciones con Lógica de Negocio Compleja y Estructuras de Control
# 
# OBJETIVO: Crear un liquidador automatizado de Impuesto a la Renta Progresivo.
# El impuesto no se cobra sobre una tasa única, sino por tramos marginales:
#   - Tramo 1: Hasta $15,000 -> Exento (0%)
#   - Tramo 2: De $15,001 a $40,000 -> 15% sobre el excedente de $15,000
#   - Tramo 3: Más de $40,000 -> 25% sobre el excedente de $40,000 + impuesto acumulado anterior.
# ------------------------------------------------------------------------------

#print("=== CLASE AVANZADA: FUNCIONES DE LIQUIDACIÓN POR TRAMOS ===\n")

def liquidar_renta_progresiva(contribuyente, renta_gravable): # Se declara la función con su respectivo nombre y con dos parametros
    """
    Calcula el impuesto sobre la renta aplicando tarifas progresivas por tramos.
    Incluye validación de tipos de datos y control de inconsistencias fiscales.
    """
    # 1. VALIDACIÓN DE ENTRADAS (Robustez del código)
    if not isinstance(renta_gravable, (int, float)): #Evalúa si el valor guardado en renta_gravable NO es un número entero (int) ni decimal (float).
        raise TypeError(f"Error en {contribuyente}: La renta gravable debe ser un número.") # Lanza el error y detiene la ejecución
        
    if renta_gravable < 0: # Verificar si es un valor negativo.
        # En el return devuelve un estado el contribuyente  y un mensaje en caso de que el valor sea negativo
        return {
            "Contribuyente": contribuyente,
            "Estado": "RECHAZADO",
            "Mensaje": "Error: Renta gravable negativa. Posible inconsistencia o fraude."
        }
    
    # 2. DEFINICIÓN DE LÍMITES Y TASAS (Parámetros fijos internos) y de acuerdo a la correspondiente normativa y Ley
    LIMITE_1 = 15000 # estos representan los topes o los umbrales
    LIMITE_2 = 40000
    TASA_TRAMO_2 = 0.15 # se define los porcentajes de impuestos
    TASA_TRAMO_3 = 0.25
    
    impuesto_total = 0.0 # se crea una variable en cero para en esta ir guardando los resultados finales
    tramo_aplicado = "" # Se crea un texto vacía que servirá para etiquetar el perfil del contribuyente según nivel de ingresos

    # 3. LÓGICA DE CÁLCULO POR TRAMOS MARGINALES 
    if renta_gravable <= LIMITE_1: # Comparar con cada límite y tramo
        # Tramo 1: Exento
        impuesto_total = 0.0 # Asigna cero al impuesto. Es el mínimo exento que protege el mínimo vital del ciudadano.
        tramo_aplicado = "Tramo 1 (Exento)"
        
    elif renta_gravable <= LIMITE_2: #Si no se cumplió lo anterior, evalúa si la renta es menor o igual a $40,000 
                                    #(es decir, está entre $15,001 y $40,000).
        
        # Tramo 2: Aplica solo sobre lo que supera el límite 1
        # Al contribuyente no se le cobra el 15% sobre todo su dinero, sino únicamente sobre lo que superó el primer tramo exento 
        excedente_t2 = renta_gravable - LIMITE_1 
        
        # Multiplica ese excedente por el 0.15 (15%) para hallar el impuesto de este tramo.
        impuesto_total = excedente_t2 * TASA_TRAMO_2
        tramo_aplicado = "Tramo 2 (15% Marginal)"
        
    else: #Si la renta no cayó en ninguno de los tramos anteriores, significa por descarte que es mayor a $40,000.
        # Tramo 3: Impuesto máximo del tramo 2 + el 25% de lo que supere el límite 2
        impuesto_max_t2 = (LIMITE_2 - LIMITE_1) * TASA_TRAMO_2  # Impuesto fijo acumulado del tramo anterior
        excedente_t3 = renta_gravable - LIMITE_2 # Calcula cuánto dinero excede el límite superior de $40,000.
        
        #Suma el impuesto acumulado del tramo anterior más el 25% (0.25) calculado estrictamente sobre este nuevo excedente. 
        impuesto_total = impuesto_max_t2 + (excedente_t3 * TASA_TRAMO_3)
        tramo_aplicado = "Tramo 3 (25% Marginal + Acumulado)"

    # 4. RETORNO DE UN EXPEDIENTE COMPLETO (Estructura de Diccionario)
    # Calcular la tasa efectiva real (impuesto / renta total) para el perfil de riesgo
    tasa_efectiva = (impuesto_total / renta_gravable * 100) if renta_gravable > 0 else 0.0
    
    # Finalmente se da instrucción de lo que se desea mostrar al final de la función empaquetado en un diccionario
    
    return {
        "Contribuyente": contribuyente,
        "Estado": "PROCESADO",
        "Renta_Gravable": round(renta_gravable, 2),
        "Impuesto_Liquidado": round(impuesto_total, 2),
        "Tasa_Efectiva": f"{round(tasa_efectiva, 2)}%",
        "Clasificacion": tramo_aplicado
    }


# ==============================================================================
# PRUEBA EN VIVO: Simulación de fiscalización con diferentes perfiles
# ==============================================================================

# Caso A: Contribuyente en tramo exento
resultado_A = liquidar_renta_progresiva("Contribuyente Pequeño S.A.", 12000)
print(f"Resultado A: {resultado_A}\n")

# Caso B: Contribuyente en tramo medio
resultado_B = liquidar_renta_progresiva("Doctora Pepito Pérez", 30000)
print(f"Resultado B:\n  -> Impuesto: ${resultado_B['Impuesto_Liquidado']}\n  -> Tasa Real: {resultado_B['Tasa_Efectiva']} ({resultado_B['Clasificacion']})\n")

# Caso C: Contribuyente de altos ingresos (Gran Contribuyente)
resultado_C = liquidar_renta_progresiva("Corporación Fantasma S.A.", 85000)
print(f"Resultado C:\n  -> Impuesto: ${resultado_C['Impuesto_Liquidado']}\n  -> Tasa Real: {resultado_C['Tasa_Efectiva']} ({resultado_C['Clasificacion']})\n")

# Caso D: Alerta de Auditoría (Datos inconsistentes/negativos)
resultado_D = liquidar_renta_progresiva("Comercializadora XX", -5000)
print(f"Resultado D (Alerta): {resultado_D}\n")
```

**3.1 Modularización**

La modularización de cálculos es una técnica de diseño que consiste en dividir un cálculo complejo en partes más pequeñas, independientes y especializadas, llamadas módulos.

En lugar de resolver un problema de estimaciones o cálculos en un solo bloque gigante de código o en una sola fórmula interminable, el **proceso se fragmenta** en pasos lógicos que se calculan por separado y luego se integran.

Un ejemplo bastante didáctico es por ejemplo el cálculo del salario neto de un empleado. El enfoque no modular es simplemente crear una sola fórmula gigantesca con cada una de las componentes y variables necesarias para la estimación.

En enfoque modular divide el problema en cuatro módulos:

* **Módulo A:** Calcula el Salario Bruto (Horas × Tarifa).
* **Módulo B:** Calcula las Deducciones de Impuestos.
* **Módulo C:** Calcula los Bonos o Incentivos.
* **Módulo Central:**  Toma los resultados de A, B y C para dar el Salario Neto (A - B + C).

En el caso de Python, es lo que se ha venido haciendo con las definiciones y creaciones de las funciones a continuación un ejemplo de una función que presenta la estructura en código de lo que es modulariar cálculos.

```python
import os
import pandas as pd
import re
import zipfile

# =============================================================================
# BLOQUE 1 - Modularización de cálculos
# =============================================================================
def calcular_impuesto_renta(ingreso_gravable: float, deducciones: float) -> float:
    """Calcula el impuesto sobre la renta aplicando una tarifa progresiva simulada."""
    base_imponible = max(0.0, ingreso_gravable - deducciones)
    
    if base_imponible <= 20000:
        impuesto = base_imponible * 0.10
    elif base_imponible <= 50000:
        impuesto = (20000 * 0.10) + ((base_imponible - 20000) * 0.20)
    else:
        impuesto = (20000 * 0.10) + (30000 * 0.20) + ((base_imponible - 50000) * 0.30)
        
    return round(impuesto, 2)

def calcular_iva(monto_base: float, tasa: float = 0.16) -> float:
    """Calcula el IVA estandarizado para auditorías de facturación."""
    if monto_base < 0:
        return 0.0
    return round(monto_base * tasa, 2)
```

Como se puede ver en la definición de la función anterior se ha generado una modularización del cálculo del Impuesto a la Renta y del Impuesto al Valor Agregado.

Definiendo en primera instancia unos tramos propuestos sobre los cuales se aplicarán el correspondiente impuesto, definiendo adicionalmente la correspondiente tarifa.

Se podría resumir la modularización como la definición de la BI, construcción de los tramos, definición de las tarifas por tramos y el cálculo del impuesto que unifica los procesos anteriores.

**3.2 Lectura y escritura de archivos ZIP/CSV**

Muchos de los contribuyentes o sistemas suelen empaquetar la información. Con zipfile y pandas podemos consolidar múltiples meses en un solo DataFrame en memoria, reduciendo líneas de código y uso de almacenamiento.

Por lo que en este bloque se va a crear funciones que sirvan para desempaquetar los archivos ZIP de tres meses con datos de los ingresos de los contribuyentes.

```python
# =============================================================================
# BLOQUE 2 - Lectura y escritura de archivos ZIP/CSV
# =============================================================================
def procesar_archivos_desde_zip(nombre_zip: str = "curso_python2/enero_marzo.zip") -> str:
    """
    Abre el archivo ZIP ubicado en la carpeta del curso, lee los tres archivos 
    internos (enero, febrero, marzo) sin extraerlos a disco y los consolida 
    en un único archivo maestro 'contribuyentes_consolidado.csv'.
    """
    lista_dataframes = []
    
    # Abrir el contenedor ZIP en la ruta especificada
    with zipfile.ZipFile(nombre_zip, 'r') as archivo_zip:
        # Listar los archivos que contiene el ZIP
        archivos_internos = archivo_zip.namelist()
        print(f"\n--- CONTENIDO DETECTADO EN EL ZIP ({nombre_zip}) ---")
        print(f"Archivos encontrados: {archivos_internos}")
        
        # Leer cada archivo CSV de forma dinámica
        for nombre_csv in archivos_internos:
            if nombre_csv.endswith('.csv'):
                # Abrir el archivo en memoria pasándolo directamente a Pandas
                with archivo_zip.open(nombre_csv) as archivo_individual:
                    # CORRECCIÓN DE SINTAXIS:
                    # 1. sep=';' -> Indica a Pandas que las columnas se dividen por punto y coma.
                    # 2. encoding='latin-1' -> Procesa correctamente acentos y eñes del estándar de Windows/Excel.
                    df_mes = pd.read_csv(archivo_individual, sep=';', encoding='latin-1')
                    
                    # Añadimos una columna para saber el origen de los datos
                    df_mes['Origen_Archivo'] = nombre_csv
                    lista_dataframes.append(df_mes)
    
    # Concatenar/unificar los tres archivos en una sola tabla maestra
    df_consolidado = pd.concat(lista_dataframes, ignore_index=True)
    
    print("\n--- DATOS UNIFICADOS DESDE EL ZIP ---")
    print(df_consolidado)
    
    # Guardar el resultado consolidado para el siguiente bloque de validación
    archivo_salida = 'contribuyentes_consolidado.csv'
    df_consolidado.to_csv(archivo_salida, index=False)
    print(f"\n[INFO] Consolidado guardado con éxito en: '{archivo_salida}'")
    return archivo_salida
```

El proceso de igual forma sigue un esquema de modularización en primera instancia se define la función encargada de la lectura masiva. Tiene por defecto la ruta de la carpeta del curso que creamos en Anaconda.

Se crea una lista vacía. Donde se va guardando la tabla de cada mes individual antes de pegarlas todas juntas.

Posteriormente y haciendo uso de la función **zipfile.ZipFile** se abre el archivo comprimido en modo lectura **('r')**. La instrucción **with** asegura que el archivo se cierre automáticamente cuando termine de usarse.

El siguiente paso es **extraer los nombres de los archivos** y asignarlos a la variable archivos_internos, esto permite una mejor lectura de cada uno de los CSV comprimidos.

A continuación, se hace uno de unos de los bucles aprendidos con el auxiliar **for**, lo que hace exactamente es iterar sobre cada archivo de la lista considerando únicamente los archivos que sean CSV **endswith** para posteriormente abrirlos el archivo CSV directamente en la memoria RAM. No lo extrae físicamente **archivo_zip.open**.

El siguiente punto es transformar el archivo de texto en un objeto DataFrame **pd.read_csv** considerando las claves de separación de los archivos así como el reconocimiento de caracteres especiales.

Para poder diferenciar de que mes vienen se añade una columna adicional **Origen_Archivo** y posteriormente se hace un append a lista vacía que se creó al inicio del proceso.

Finalmente, se consolida los archivos **pd.concat** y se guarda en un csv consolidado.

**3.3 Validación de datos**

Es un proceso de comprobar que la información recopilada cumple con **reglas, formatos y estándares establecidos** antes de ser analizada o ingresada en un sistema. Para la presente capacitación se ha desarrollado unas funciones que permita validar el formato alfanumérico de uno de los campos de los csv anteriormente cargados.

```python
# =============================================================================
# BLOQUE 3  - Validación de datos
# =============================================================================
def validar_identificador_fiscal(id_fiscal: str) -> bool:
    """Valida si el formato cumple con un estándar alfanumérico fiscal."""
    # [A-Z]{3,4}: Busca entre 3 y 4 letras mayúsculas consecutivas (de la A a la Z).
    # d{6}: Busca exactamente 6 dígitos numéricos (del 0 al 9).
    # [A-Z0-9]{3}: Busca exactamente 3 caracteres alfanuméricos (pueden ser letras mayúsculas o números).
    patron = r'^[A-Z]{3,4}\d{6}[A-Z0-9]{3}$'
    return bool(re.match(patron, str(id_fiscal)))

def ejecutar_auditoria_automatizada(archivo_csv: str):
    """Aplica las funciones anteriores para auditar el archivo consolidado."""
    if not os.path.exists(archivo_csv):
        print(f"Error: El archivo {archivo_csv} no existe.")
        return

    df = pd.read_csv(archivo_csv)
    
    # 1. Aplicar modularización (Bloque 1)
    
    # df.apply(..., axis=1): Le dice a Pandas que recorra el DataFrame fila por fila (horizontalmente), 
    # en lugar de columna por columna.lambda row:: Es una forma rápida de decir: "Para cada fila que vayas recorriendo, 
    # llámala temporalmente row y ejecuta lo siguiente".row['Ingresos'] y row['Deducciones']: 
    # Al tener la fila completa en la variable row, puedes acceder a los valores específicos de las columnas Ingresos y Deducciones de esa fila en particular.calcular_impuesto_renta(...): Es la función que toma esos dos valores individuales, hace el cálculo y devuelve un único resultado para esa fila.
    
    df['Impuesto_Calculado'] = df.apply(
        lambda row: calcular_impuesto_renta(row['Ingresos'], row['Deducciones']), axis=1
    )
    
    # 2. Validación de Datos (Bloque 4)
    df['Fiscal_Valido'] = df['Identificador_Fiscal'].apply(validar_identificador_fiscal)
    df['Alerta_Ingreso_Negativo'] = df['Ingresos'] < 0
    
    print("\n--- INFORME DE AUDITORÍA Y VALIDACIÓN (BASE CONSOLIDADA) ---")
    print(df[['ID_Contribuyente', 'Origen_Archivo', 'Impuesto_Calculado', 'Fiscal_Valido', 'Alerta_Ingreso_Negativo']])
    
    # Exportar hallazgos de fiscalización
    df_alertas = df[(df['Fiscal_Valido'] == False) | (df['Alerta_Ingreso_Negativo'] == True)]
    df_alertas.to_csv('casos_para_revision.csv', index=False)
    print("\n[ALERTA] Se ha generado 'casos_para_revision.csv' con las anomalías detectadas en los 3 meses.")
```

Como primer punto se crea una funcion **validar_identificador_fiscal** donde se menciona al campo el cual va a ser analizado, y el tipo de formato que debe tener esto mediante una evaluación mediante una **Expresión Regular (RegEx)** para fiscalizar formatos de texto mediante **re.match** se compara con id del contribuyente.

A continuación se genera otra función que procesará la auditoría sobre cualquier archivo CSV que le pasemos y cuyos paramentros serán precisamente los archivos csv. Se hace una validación de que si el archivo consolidado no existe, detiene la función (return) antes de que Python arroje un error crítico en pantalla.

Una vez creadas las funciones validadoras, se proceden a consolidar a la memoria los archivos y se ejecuta el procesamiento fila por fila **df.apply** de los cálculos del impuesto a la renta según el bloque1 **lambda row:**.

Luego se aplica la función **validar_identificador_fiscal** a toda la columna de identificadores fiscales, creando una nueva columna con marcas True (ID válido) o False (ID erróneo/evasor).

Adicionalmente, se realiza una validación de si los ingresos presentan valores negativos y se muestra un reporte con el resumen seleccionando únicamente las columnas de interés para el análisis, ocultando datos de nombres para proteger el análisis confidencial.

Para finalizar se realiza un filtrado y se exporta lo hallazgos a CSV, de manera concreta las filas que fallaron.

**5.4 Buenas prácticas y manejo de errores en Python**

A continuación, encontrarás una guía estructurada con los errores más comunes y cómo solucionarlos aplicando la filosofía de Python y el estándar **PEP 8**.

El **estándar PEP 8** (Python Enhancement Proposal 8) es la guía de estilo oficial para código Python, creada en 2001 por Guido van Rossum (el creador del lenguaje) junto a Barry Warsaw y Nick Coghlan

**¿Para qué sirve?**

* **Legibilidad:** Busca que el código sea limpio, coherente y fácil de leer para cualquier programador.

* **Mantenimiento:** Facilita corregir errores o actualizar proyectos a largo plazo.

* **Colaboración:** Permite que equipos de trabajo entiendan rápidamente el código ajeno porque todos siguen las mismas reglas.

**Reglas principales de PEP 8**

* **Nombres de variables y funciones:** Se usa snake_case (letras minúsculas separadas por guiones bajos).
* **Nombres de clases:** Se usa CapWords (la primera letra de cada palabra en mayúscula).
* **Identación:** Se deben usar 4 espacios por nivel de anidación (evitando usar tabulaciones sueltas).
* **Longitud de línea:** Se recomienda un máximo de 79 caracteres por línea para facilitar la visualización.
* **Espacios en blanco:** Evitar espacios innecesarios alrededor de paréntesis, corchetes o antes de comas.

**Buenas prácticas**

**Concatenación de Cadenas (Strings)**

Modificar strings en bucles utilizando el operador + es sumamente ineficiente porque las cadenas en Python son inmutables; cada suma crea un nuevo objeto en memoria.

```python
palabras = ["Python", "es", "un", "gran", "lenguaje"]
frase = ""
for p in palabras:
    frase += p + " "  # Ineficiente en bucles grandes
```

**Comprensión de Listas (List Comprehensions)**

Crear listas vacías para luego llenarlas con un bucle for básico añade líneas de código innecesarias.

```python
numeros = [1, 2, 3, 4, 5]
cuadrados = []
for n in numeros:
    if n % 2 == 0:
        cuadrados.append(n ** 2)
```

```python
numeros = [1, 2, 3, 4, 5]
cuadrados = [n ** 2 for n in numeros if n % 2 == 0]
```

El código se organiza en 4 partes:

## **1. Configuración del cliente y del contexto legal**

**texto_reforma_tributaria**: Es la base de conocimientos. Contiene las reglas del juego (las leyes reales sobre el IVA y créditos tributarios en software) que la IA debe respetar.

## **2. Definición de roles y prompts**

Los **prompts** es la instrucción, pregunta o texto que le das a una Inteligencia Artificial para que te devuelva una respuesta, es decir, es una especie de **orden exacta** que se le da para que haga un determinado trabajo.

Entre los principales componentes se tiene:

* **El Rol (Actúa como):** Le dice a la IA qué identidad debe adoptar (ej. "Actúa como un programador experto en Python").
* **El Contexto:** Le da antecedentes a la IA para que entienda la situación (ej. "Estoy creando una app móvil para personas de la tercera edad...").
* **La Tarea (La instrucción principal):** Qué es exactamente lo que debe hacer (ej. "Escribe una función para recuperar contraseñas").
* **Las Restricciones o Formato:** Cómo quieres que te entregue el resultado (ej. "Explícalo sin usar tecnicismos, en menos de 200 palabras y usa viñetas").

En el caso del ejemplo planteado se encuentran definidos en dos prompts uno de sistema y otro de usuario.

* **prompt_sistema:** Le da la identidad a la IA ("Eres un auditor fiscal senior...") y le exige que se limite estrictamente al texto proporcionado para evitar "alucinaciones" o inventos.

* **prompt_usuario:** Une la ley con el caso específico del cliente que se quiere analizar.

## **3. Estrategia de Fallback (Tolerancia a fallos)**

Esta es la parte más avanzada del script:

* **modelos_a_probar = [...]:** Define una lista de prioridades. Primero intentará usar gemini-3.6-flash.
* **El bucle for y try/except APIError:** Si Google devuelve un Error 503 (servidor saturado), el código no se detiene; muestra una advertencia, espera 1 segundo (time.sleep(1)) y reintenta la misma consulta automáticamente con el siguiente modelo en la lista (gemini-3.5-flash-lite).

* **return "❌ Error...":** Si absolutamente todos los modelos fallan, el código tiene una salida elegante en lugar de cerrarse abruptamente con un error de sistema.

## **4. Ejecución (Demostración)**

El script evalúa el caso: "Empresa de software exporta el 100% de servicios a Europa".

La IA procesará esto frente al Artículo 45 y debería dictaminar que la empresa tiene derecho a un crédito tributario del 5% sobre sus adquisiciones de capital.

```python
import os
import time
from google import genai
from google.genai import types
from google.genai.errors import APIError  # Para atrapar errores del servidor de Google

client = genai.Client(api_key="AQ.Ab8RN6Iou7ZLpyW6ZWIXENu01a5ucmzHe-7Wh2vttp300pUHtw")

texto_reforma_tributaria = """
Exenciones al Impuesto al Valor Agregado (IVA).
Estarán exentos del pago del IVA los servicios digitales de educación formal prestados 
por instituciones validadas por el Ministerio. Sin embargo, las plataformas de streaming 
y entretenimiento masivo mantendrán una tarifa general del 15%. 
Las empresas de desarrollo de software local que exporten el 100% de sus servicios 
tendrán derecho a un crédito tributario del 5% sobre sus adquisiciones de capital.
"""

def analizar_caso_fiscal(caso_cliente: str) -> str:
    prompt_sistema = (
        "Eres un auditor fiscal senior experto en legislación tributaria. "
        "Analiza el caso del cliente utilizando estrictamente el texto legal proporcionado."
    )
    
    prompt_usuario = f"Normativa Legal:\n{texto_reforma_tributaria}\n\nCaso:\n{caso_cliente}"
    
    # Lista de modelos a intentar en orden de preferencia (Estrategia Fallback)
    modelos_a_probar = ['gemini-3.6-flash', 'gemini-3.5-flash-lite'] #
    
    for modelo in modelos_a_probar:
        print(f"📡 Intentando análisis con el modelo: {modelo}...")
        try:
            response = client.models.generate_content(
                model=modelo,
                contents=prompt_usuario,
                config=types.GenerateContentConfig(
                    system_instruction=prompt_sistema,
                    temperature=0.2,
                ),
            )
            return response.text  # Si funciona, rompe el bucle y devuelve el resultado
            
        except APIError as e:
            # Si el error es una saturación del servidor (503), avisamos y pasamos al siguiente
            if e.code == 503:
                print(f"⚠️ El modelo {modelo} está saturado (Error 503). Probando alternativa...")
                time.sleep(1) # Pausa técnica de cortesía
                continue
            else:
                # Si es otro tipo de error (ej. permisos), lo lanzamos de inmediato
                raise e
                
    # Si todo falla, activamos el plan de contingencia local para no congelar la app
    return "❌ Error del servidor de Google: Todos los modelos remotos están temporalmente saturados."

# --- DEMOSTRACIÓN ---
if __name__ == "__main__":
    caso_auditoria = "Empresa de software exporta el 100% de servicios a Europa."
    resultado = analizar_caso_fiscal(caso_auditoria)
    print("\n📝 DICTAMEN FINAL:")
    print(resultado)
```

## El Bloque de Control de Errores (try-except)

**try:** Le dice a Python: "Intenta ejecutar las siguientes líneas. Si algo sale mal, no te detengas ni rompas el programa, ve al bloque except".

## Captura y Manejo del Error 503 (Saturación)

**except APIError as e:** Si la API responde con un error, lo capturamos bajo la variable e para analizar sus propiedades.

**if e.code == 503:** Evaluamos si el código de error HTTP es un 503 (Servidor Saturado).

**time.sleep(1) y continue:** Si es un 503, esperamos un segundo para dar respiro y usamos continue para saltar de inmediato a la siguiente iteración del bucle for, probando el siguiente modelo de la lista.

**else: raise e:** Si el error no es un 503 (por ejemplo, si es un error 400 por una API Key inválida), no tiene sentido probar otro modelo. Usamos raise e para relanzar el error y detener el programa, ya que es un fallo estructural del código o credenciales.

## Plan de Contingencia Final y Ejecución

**return "❌ Error...":** Esta línea solo se ejecuta si el bucle for termina por completo y todos los modelos de la lista fallaron con un error 503. Es la última red de seguridad para que la aplicación devuelva un mensaje controlado en lugar de colapsar.
